### step-1 
import libraries and load dataset


```python
import pandas as pd
import numpy as np

df = pd.read_csv('raw_dataset.csv')

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)


df.head()
```

### Step 2 — Inspect the dataset

```python
# Check column names
print(df.columns.tolist())

# Check data types and non-null values
df.info()

# Count missing values
print(df.isnull().sum())

# Count completely duplicated rows
print("Duplicate rows:", df.duplicated().sum())

# Summary of numerical columns
display(df.describe())
```

### step-3 Identify problem
 


```python
# Find blank or whitespace-only cells
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].replace(r"^\s*$", np.nan, regex=True)

print(df.isnull().sum())
```

### Step -4 Handle Missing values

```python
 # Investigate missing values in percentages
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

display(missing_summary)
```

```python
# Display the first 10 rows with missing values
display(df[df.isnull().any(axis=1)].head(10))
```

### Step-5 checking the missing values before fixing them

```python
# Check whether missing values occur together
missing_rows = df[df.isnull().any(axis=1)]

display(
    missing_rows[
        ["Item", "Category", "Price Per Unit",
         "Quantity", "Total Spent", "Discount Applied"]
    ].head(20)
)
```

```python
# Now investigate Discount Applied seprately
print("Discount Applied values:")
print(df["Discount Applied"].value_counts(dropna=False))

print("\nData type:")
print(df["Discount Applied"].dtype)
```

### step -6 check incorrect and inconsistent values

```python
# Check numeric values
print("Price Per Unit:")
print(df["Price Per Unit"].describe())

print("\nQuantity:")
print(df["Quantity"].describe())

print("\nTotal Spent:")
print(df["Total Spent"].describe())

# Check for zero or negative values
print(
    "Non-positive prices:",
    (df["Price Per Unit"] <= 0).sum()
)

print(
    "Non-positive quantities:",
    (df["Quantity"] <= 0).sum()
)
```

```python
#Check inconsistent text
for col in [
    "Category", "Payment Method", "Location", "Item"
]:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False).head(20))
```

```python
#Check transaction dates
print(df["Transaction Date"].head(10))

parsed_dates = pd.to_datetime(
    df["Transaction Date"],
    errors="coerce"
)

print("Dates that cannot be parsed:", parsed_dates.isna().sum())
```

### Step 7 — Check whether the totals are correct

```python
check = df.dropna(
    subset=["Price Per Unit", "Quantity", "Total Spent"]
).copy()

check["Calculated Total"] = (
    check["Price Per Unit"] * check["Quantity"]
)

check["Difference"] = (
    check["Total Spent"] - check["Calculated Total"]
).round(2)

display(
    check.loc[
        check["Difference"].abs() > 0.01,
        ["Price Per Unit", "Quantity",
         "Total Spent", "Calculated Total", "Difference"]
    ].head(20)
)
```

```python
 # Investigate missing values in percentages
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

display(missing_summary)
```

```python
# Preserve the original data in memory
df_raw = df.copy(deep=True)

# Work on a separate copy
df_clean = df.copy(deep=True)

print("Original shape:", df_raw.shape)
print("Working copy shape:", df_clean.shape)
```

```python
text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df_clean[col] = df_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Unnecessary surrounding spaces removed.")
```

```python
df_clean["Transaction Date Parsed"] = pd.to_datetime(
    df_clean["Transaction Date"],
    errors="coerce"
)

invalid_dates = (
    df_clean["Transaction Date"].notna()
    & df_clean["Transaction Date Parsed"].isna()
)

print("Invalid dates:", invalid_dates.sum())
display(df_clean.loc[invalid_dates, ["Transaction Date"]].head(20))
```

```python
# Run only after verifying the date format
if not invalid_dates.any():
    df_clean["Transaction Date"] = df_clean.pop(
        "Transaction Date Parsed"
    )
```

```python
missing_summary = pd.DataFrame({
    "Missing Count": df_clean.isnull().sum(),
    "Missing Percentage": (
        df_clean.isnull().mean() * 100
    ).round(2)
})

display(missing_summary)
```

```python
display(
    df_clean.loc[
        df_clean.isnull().any(axis=1),
        [
            "Transaction ID",
            "Category",
            "Item",
            "Price Per Unit",
            "Quantity",
            "Total Spent",
            "Discount Applied"
        ]
    ].head(20)
)
```

```python
print(
    df_clean["Discount Applied"]
    .value_counts(dropna=False)
)
```

```python
check = df_clean.dropna(
    subset=["Price Per Unit", "Quantity", "Total Spent"]
).copy()

check["Calculated Total"] = (
    check["Price Per Unit"] * check["Quantity"]
).round(2)

check["Difference"] = (
    check["Total Spent"] - check["Calculated Total"]
).round(2)

inconsistent_totals = check[
    check["Difference"].abs() > 0.01
]

print("Records checked:", len(check))
print("Inconsistent totals:", len(inconsistent_totals))

display(
    inconsistent_totals[
        [
            "Transaction ID",
            "Price Per Unit",
            "Quantity",
            "Total Spent",
            "Calculated Total",
            "Difference"
        ]
    ].head(30)
)
```

```python
print("Exact duplicate rows:", df_clean.duplicated().sum())

for col in ["Price Per Unit", "Quantity", "Total Spent"]:
    print(f"\n{col}")
    print(df_clean[col].describe())
    print("Negative values:", (df_clean[col] < 0).sum())
    print("Zero values:", (df_clean[col] == 0).sum())
```

```python
df_clean.to_csv(
    "cleaning_review.csv",
    index=False
)

print("Review copy saved.")
print("Original rows:", len(df_raw))
print("Review rows:", len(df_clean))
```
